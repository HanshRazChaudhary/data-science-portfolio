### Advance Operation With Business Data

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("==== Zomato Sales Analysis ====")
print("\n Sales Data Shape:", sales_data.shape)
print("\n Sample Data Of First 3 Resturant: \n", sales_data[:3])
```

```python
# Total Sales Per Year:
print("Sum Of Year Sales Columns Wise:",np.sum(sales_data[:, 1:], axis = 0))
print("Sum Of Year Sales Rows Wise:",np.sum(sales_data[:, 1:], axis = 1))
print("Sum Of Yearly Sales:",np.sum(np.sum(sales_data[:, 1:], axis = 0)))
```

```python
# Minmum Sales Per Year:
print("Minmum Sales Of Resturant (Row Wise):", np.min(sales_data[:, 1:], axis = 1))
```

```python
# Maximum Sales Of The Year:
print("Maximum Sales Of The Year:", np.max(sales_data[:, 1:], axis = 0))
```

```python
# Average Sales Per Restutant:
print("Avgrage Sale Per Resturant As Row Wise:", np.mean(sales_data[:, 1:], axis = 1))
```

```python
# Cummlative Sales Per Resturant:
cumsum = np.cumsum(sales_data[:, 1:], axis = 1)
print("Cummalative Sales Per Year Of Resturant:\n", cumsum)
```

```python
# Plotting Cummulative Data Mean:
plt.figure(figsize = (10, 6))
plt.plot(np.mean(cumsum, axis = 0))
plt.title("Avgrage Cummulative Sales Across Year")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

### Arthematic Operation On Vector

```python
# Arhematic Operation On Vector :
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])

print("Vetor Additon:\n", vector1 + vector2)
print("\nVetor Multiplication:\n", vector1 * vector2)
print("\nDot Product:\n", np.dot(vector1, vector2))
```

```python
# Finding Angle:
angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("Angle:", angle)
```
