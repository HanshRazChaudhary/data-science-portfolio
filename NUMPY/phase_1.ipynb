## LEARNING NUMPY

# Importing Numpy

```python
# Importing Numpy:
import numpy as np
```

### Creating Array From List

```python
arr_1d = np.array([1,2,3])
print('1D Array:', arr_1d)

arr_2d = np.array([[1,2,3], [4,5,6]])
print('2D Array:\n', arr_2d)
```

### Python List VS NumPy Array

```python
py_list = [1,2,3]
print("Python List Multiplication: ", py_list * 2)

np_array = np.array([4,5,6]) # Element Wise Multiplication
print("Python Array [NumPy] Multiplication: ", np_array * 2)
```

```python
# Working With Time And Numpy:
import time
# Python List
start = time.time()
py_list1 = [x * 2 for x in range(1000000)]
print("\n List Operation Time:", time.time() - start)

# NumPy Or Python Array
start = time.time()
np_array1 = np.arange(1000000) * 2
print("\n NumPy Operation Time:", time.time() - start)
```

### Creating Array From Scratch

```python
# Creating Array Of Zeros:
zeros = np.zeros((3, 4))
print("Zero Array: \n", zeros)

# Creating Array For Ones:
ones = np.ones((3, 4))
print("One Array: \n", ones)

# Creating Constant Array:
full = np.full((3, 4), 7)
print("Full Array: \n", full)

# Creating Random Array:
random = np.random.random((3, 4))
print("Random Array: \n", random)

# Creation Sequence Of Array Using ARRAnge:
seq = np.arange(0, 10, 2)    # (start, stop, step)
print("Sequence Array: \n", seq)
```

### Vector, Matrix And Tensor

```python
# Crearing Vector: (1 X N) - 1D
vector = np.array([1, 2, 3])
print("Vector: \n", vector)

# Creating Matrix: (2 X N) - 2D
matrix = np.array([[1, 2, 3],
                  [4, 5, 6]])
print("Matrix: \n", matrix)

# Creating Tensor (N X N) - ND
tensor = np.array([[[1, 2], [3, 4],
                    [5, 6], [7, 8]]])
print("Tensor: \n", tensor)
```

### Array Properties

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print('Shape: ', arr.shape)
print('Size: ', arr.size)
print('Dimension: ', arr.ndim)
print('DType: ', arr.dtype)
```

### Reshaping Array:

```python
arr1 = np.arange(12)
print("Original Array:", arr1)

# Reshaping Our Original Array:
reshaped = arr1.reshape((3, 4))
print("Reshaped Array: \n", reshaped)

# Flatten Our Reshaped Array:
flattened = reshaped.flatten()
print("Flatten Array:", flattened)

# Raveled Our Reshaped Array:
# RAVEL [Return View, Instead Of Copy]
ravaled = reshaped.ravel()
print("Ravaled Array:", ravaled)

# Transposed Array:
transposed = reshaped.T
print("Transposed Array: \n", transposed)
```
