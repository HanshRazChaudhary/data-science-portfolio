```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Creating Different Array:
array1 = np.array([[1, 2, 3], [4, 5, 6]])
array2 = np.random.rand(3, 3)
array3 = np.zeros((4, 4))
```

```python
# Saving That Three Array [In .npy Format]:
np.save('array1', array1)
np.save('array2', array2)
np.save('array3', array3)
```

```python
# Printing Array Which We Saved:
loaded_array1 = np.load('array1.npy')
print("First Array: \n", loaded_array1)
loaded_array2 = np.load('array2.npy')
print("Second Array: \n", loaded_array2)
loaded_array3 = np.load('array3.npy')
print("Third Array: \n", loaded_array3)
```

```python
# Loading Our NumPy Logo:
try:
    # Loading Our NumPy Logo:
    num_logo = np.load('numpy-logo.npy')

    # Displaying Our Logo:
    plt.Figure(figsize = (10, 5))
    plt.subplot(121)
    plt.imshow(num_logo)
    plt.title("NumPy Logo")
    plt.grid(False)
    
    # Making Our Logo Dark:
    dark_log = 1 - num_logo
    plt.subplot(122)
    plt.imshow(dark_log)
    plt.title("NumPy Dark Logo")
    plt.grid(False)

except FileNotFoundError:
    print("NumPy Logo Not Found!")
```
