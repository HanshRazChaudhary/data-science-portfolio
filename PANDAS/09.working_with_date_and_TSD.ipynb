```python
# Filter Data:
# Getting Data After Yesr 2020:
filt = (df['Date'] >= '2020')
df[filt]
```

```python
# Getting From Jan 1 To Dec 31 Of 2019:
filt2 = (df['Date'] >= '2019') & (df['Date'] < '2020')
df[filt2]
```

```python
# Getting Same Result With Different Approach:
filt3 = (df['Date'] >= pd.to_datetime('2019-01-01')) & (df['Date'] < pd.to_datetime('2020-01-01'))
df[filt3]       
```
