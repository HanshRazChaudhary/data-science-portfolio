```python
import pandas as pd
import numpy as np
```

```python
people = {
    'first' : ['Ram', 'Shyam', 'Binod', 'Hari', np.nan, None, 'NA'],
    'last' : ['Das', 'Bahadur', 'Chaudhary', 'Rai', np.nan, np.nan, 'Missing'],
    'email' : ['ramdas@gmail.com', 'shyambdr@gmail.com', 'binodchy@gmail.com', None, np.nan, 'anyone@gmail.com','NA'],
    'age' : ['35',  '45', '36', '43', None, None, 'Missing']
}
```

```python
df = pd.DataFrame(people)

# Modifying Our DataFrame With The Help Of NumPy:
df.replace('NA', np.nan, inplace = True)
df.replace('Missing', np.nan, inplace = True)
```

```python
df
```

```python
# Dropping Rows Which Have Missing Value:
df.dropna()
```

```python
# How Does This DROPNA() Works:
# If Row Have Any Missing Value Than It Would Auto Drop That Row If All Columns Have Missing  Value Than It Would Not:
df.dropna(axis = 'index', how = 'any')
```

```python
# If We Change Above Code Here:
df.dropna(axis = 'index', how = 'all')
# Than It Would Drop Rows Which All Column Have Missing Value:
```

```python
# If We Set INDEX = 'COLUMNS' and HOW = 'ALL' Then We Get Our Orginal DataFrame:
df.dropna(axis = 'columns', how = 'all')
```

```python
# If We Set INDEX = 'COLUMNS' and HOW = 'ANY' Then We Get EMPTY DataFrame:
df.dropna(axis = 'columns', how = 'any')
```

```python
# Using Subset To Get Values Only If email Column Have NaN Value:
df.dropna(axis = 'index', how = 'any', subset = ['email'])
```

```python
# To Get Values Columns In Subset Must Have Missing Value:
df.dropna(axis = 'index', how = 'all', subset = ['last', 'email'])
```

```python
# To Check How Many Values Are NaN:
df.isna()
```

```python
df.fillna('MISSING')
```

```python
# To Get DataTypes Of Columns In Our DataFrame:
df.dtypes
```

```python
# To Get Type Of Our np.nan
type(np.nan)
```

```python
# Changing Our 'AGE' Data Types/Casting Data Types As Requirement: 
df['age'] = df['age'].astype(float)
```
