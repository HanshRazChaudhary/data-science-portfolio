## Four Methods To Upadte Values
### 1. apply
### 2. map
### 3. applymap
### 4. replace

```python
df
```

```python
# Using Apply To Get Len Of E-Mail:
df['email'].apply(len)
```

```python
# Creating Function Which Return Email In Upper Case:
def upper_case_mail(email):
    return email.upper()
```

```python
df['email'].apply(upper_case_mail)
```

```python
# Setting Value From Function In DataFrame:
df['email'] = df['email'].apply(upper_case_mail)
```

```python
df
```

```python
# Setting Lower Case To Email In DataFrame Using Lambda Function:
df['email'] = df['email'].apply(lambda x : x.lower())
```

```python
df
```

```python
# It Gives Us The Rows In That Give Data Along With Column Name:
# It Count Value Columns Wise
df.apply(len)
```

```python
# Setting Axis To Get Values:
# It Count Value Rows Wise
df.apply(len, axis = 'columns')
```

```python
# Getting Value 'How Much Entry or Data Is In Email Column in Give DataFrame':
len(df['email'])
```

```python
# In Series DataFrame This Give In Columns Which Alphabetical Value Comes First:
# In Numerical Data It Gives Us Minimum Value:
df.apply(pd.Series.min)
```

```python
# Lambda Function Use Only In Series:
df.apply(lambda x : x.min())
```

```python
df.map(len)
```

```python
df.applymap(len)
```

```python
# Using map Instead Of applymap Method --> Cause ApplyMap Method Has Been Deprecated:
df.map(str.lower)
```

```python
df
```

```python
# Changing Particular Value Using Map Method:
df['first'].map({'Ram': 'Laxaman', 'Shyam': 'Naman'})
```

```python
df
```

```python
# Changing Values Usinf Replace Method:
# But This Does Not Change Value in Our DataFrame:
df['first'].replace({'Ram': 'Laxaman', 'Shyam': 'Naman'})
```

```python
# To Change Take Place In DataFrame We Have To Use:
df['first'] = df['first'].replace({'Ram' : 'Laxaman', 'Shyam' : 'Naman'})
```
