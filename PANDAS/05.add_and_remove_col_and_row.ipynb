```python
people = {
    'first' : ['Ram', 'Shyam', 'Binod'],
    'last' : ['Das', 'Bahadur', 'Chaudhary'],
    'email' : ['ramdas@gmail.com', 'shyambdr@gmail.com', 'binodchy@gmail.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['first'] + ' ' + df['last']
```

```python
# Adding New Columns In Our DataFrame:
df['full_name'] = df['first'] + ' ' + df['last']
```

```python
df
```

```python
# Removing Columns Using Drop Method:
df.drop(columns = ['first', 'last'], inplace = True)
```

```python
df
```

```python
# Spliting The Full Name Columns:
df['full_name'].str.split(' ')
```

```python
# Spliting Permanently:
df['full_name'].str.split(' ', expand = True)
```

```python
# Assing Columns To Our Splited Value:
df[['first', 'last']] = df['full_name'].str.split(' ', expand = True)
```

```python
df
```

```python
# Adding New Row In Our Existing DataFrame:
# New row as a DataFrame
new_row = pd.DataFrame([{'first': 'Mohan'}])

# Concatenate
df = pd.concat([df, new_row], ignore_index=True)

# --------------- OR ------------------------
# df = pd.concat([df, pd.DataFrame([{'first': 'Mohan'}])], ignore_index=True)
```

```python
df
```

```python
# Creating Another DataFrame:
people = {
    'first' : ['Tony', 'Steve'],
    'last' : ['Stark', 'Rogers'],
    'email' : ['IronMan@avgen.com', 'CapatinAmerica@avgen.com']
}

df2 = pd.DataFrame(people)
```

```python
df2
```

```python
# Making One DataFrame Of Two DataFrame:
df = pd.concat([df, df2], ignore_index = True)
```

```python
df
```

```python
# Deleting Row:
df.drop(index = 5, inplace = True)
```

```python
df
```

```python
# Deeting Row Using Conditional:
filt = df['last'] == 'Chaudhary'
df.drop(index = df[filt].index)
```
