```python
# Using Filter Method In Group US To Find How Many Use Python Language:
filt = df['Country'] == 'United States'
df.loc[filt]['LanguageWorkedWith'].str.contains('Python')
```

```python
# Using Filter Method In Group US To Find How Many Use Python Language:
filt = df['Country'] == 'United States'
# To Get Solid Value We Use SUM Method:
df.loc[filt]['LanguageWorkedWith'].str.contains('Python').sum()
```

```python
# In Group Of Country We Are Using String Function To Get How Many Use Python:
        # This Will Give You 'AttributeError'
# country_grp['LanguageWorkedWith'].str.contains('Python').sum()
```

```python
# Fixing Above Code Using APPLY Method And LAMBDA Function:
country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
```

```python
# Fixing Above Code Using APPLY Method And LAMBDA Function For Specific Country:
country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum()).loc['India']
```

```python
# Practice Time:
# What Percentage Of People Know Python?
python_pct = country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python', na = False).sum()).value_counts(normalize = True)
python_pct
```

```python
# Corey Solution's:
country_respondent = df['Country'].value_counts()
country_respondent
```

```python
# Getting How Many User Know Python In Each Country:
country_knows_python = country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
country_knows_python
```

```python
# Using Concat Function:
python_df = pd.concat([country_respondent, country_knows_python],axis = 'columns', sort = False)
python_df
```

```python
# Changing Columns Name:
python_df.rename(columns = {'count' : 'NumRespondent', 'LanguageWorkedWith' : 'NumKnowsPython'}, inplace = True)
```

```python
python_df
```

```python
# Calculating Percentage:
python_df['PctKnowsPython'] = (python_df['NumKnowsPython']/python_df['NumRespondent']) * 100
python_df
```

```python
# Sorting Value:
python_df.sort_values(by = 'PctKnowsPython', ascending = False, inplace = True)
```
